```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Membaca dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# 2. Pembersihan data (harga, jumlah negatif, dan format tanggal)
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)
df.loc[df['jumlah'] < 0, 'jumlah'] = df['jumlah'].abs()
df['tanggal'] = pd.to_datetime(df['tanggal'], format='mixed')

# 3. Ekstrak tahun, bulan, dan hitung pendapatan bersih
df['tahun'] = df['tanggal'].dt.year
df['bulan'] = df['tanggal'].dt.month
df['pendapatan'] = df['jumlah'] * df['harga_satuan'] * (1 - df['diskon_persen'] / 100)

# 4. Filter khusus data tahun 2025
df_2025 = df[df['tahun'] == 2025]
df_2025_selesai = df_2025[df_2025['status'] == 'SELESAI']

print("Data tahun 2025 siap divisualisasikan!")
```

```python
# Membuat kanvas 2x2 (4 bagian visualisasi)
fig, axes = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle('Dashboard Analisis Transaksi POS Tahun 2025', fontsize=16, fontweight='bold')

# --- BAGIAN 1: Penjualan per Bulan Tahun 2025 (Line Chart) ---
penjualan_bulanan = df_2025_selesai.groupby('bulan')['pendapatan'].sum()
axes[0, 0].plot(penjualan_bulanan.index, penjualan_bulanan.values, marker='o', color='#2b5c8f', linewidth=2)
axes[0, 0].set_title('1. Tren Penjualan per Bulan (2025)', fontweight='bold')
axes[0, 0].set_xlabel('Bulan (1 = Jan, 12 = Des)')
axes[0, 0].set_ylabel('Total Pendapatan (Rp)')
axes[0, 0].set_xticks(range(1, 13))
axes[0, 0].grid(True, linestyle='--', alpha=0.6)

# --- BAGIAN 2: 10 Produk Terlaris Tahun 2025 (Horizontal Bar Chart) ---
top_10_produk = df_2025_selesai.groupby('kode_produk')['jumlah'].sum().nlargest(10).sort_values()
axes[0, 1].barh(top_10_produk.index, top_10_produk.values, color='#3a86ff')
axes[0, 1].set_title('2. Top 10 Produk Terlaris (Unit Terjual - 2025)', fontweight='bold')
axes[0, 1].set_xlabel('Total Unit Terjual')
axes[0, 1].set_ylabel('Kode Produk')
axes[0, 1].grid(axis='x', linestyle='--', alpha=0.6)

# --- BAGIAN 3: Proporsi Transaksi Berhasil vs Gagal/Cancel (Pie Chart) ---
# Mengelompokkan status: SELESAI vs GAGAL/CANCEL (BATAL + RETUR)
status_kategori = df_2025['status'].apply(lambda x: 'Berhasil (SELESAI)' if x == 'SELESAI' else 'Gagal / Cancel (BATAL & RETUR)')
proporsi_status = status_kategori.value_counts()
axes[1, 0].pie(
    proporsi_status.values, 
    labels=proporsi_status.index, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=['#4caf50', '#f44336'],
    explode=(0, 0.1)
)
axes[1, 0].set_title('3. Proporsi Status Transaksi (2025)', fontweight='bold')

# --- BAGIAN 4: Hubungan Jumlah Beli dengan Diskon (Scatter Plot) ---
axes[1, 1].scatter(df_2025['jumlah'], df_2025['diskon_persen'], alpha=0.5, color='#ff9f1c', edgecolors='k', linewidth=0.5)
axes[1, 1].set_title('4. Hubungan Jumlah Pembelian vs Diskon (%)', fontweight='bold')
axes[1, 1].set_xlabel('Jumlah Pembelian (Unit)')
axes[1, 1].set_ylabel('Diskon (%)')
axes[1, 1].grid(True, linestyle='--', alpha=0.6)

# Merapikan tata letak agar tidak saling bertumpuk
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
