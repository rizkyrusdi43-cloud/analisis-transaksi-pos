```python
import pandas as pd

# Membaca dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# 1. Pembersihan Format Harga (ubah koma jadi titik, konversi ke float)
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)

# 2. Penanganan Kuantitas Negatif (ubah nilai < 0 menjadi positif dengan .abs())
df.loc[df['jumlah'] < 0, 'jumlah'] = df['jumlah'].abs()

# 3. Imputasi Nilai Kosong Kasir dengan 'K_UNKNOWN'
df['kode_kasir'] = df['kode_kasir'].fillna('K_UNKNOWN')

# 4. Seleksi Posisi Slicing (tampilkan id_transaksi, tanggal, kode_gerai, jumlah untuk 5 baris pertama)
print("--- TASK 1.4: Seleksi 5 Baris Pertama ---")
display(df.iloc[:5, df.columns.isin(['id_transaksi', 'tanggal', 'kode_gerai', 'jumlah'])])
```

```python
# 1. Perhitungan Omzet Kotor
df['Total_Kotor'] = df['harga_satuan'] * df['jumlah']

# 2. Perhitungan Diskon (diskon_persen dibagi 100)
df['Nominal_Diskon'] = df['Total_Kotor'] * (df['diskon_persen'] / 100)

# 3. Perhitungan Omzet Bersih & Pajak (Rumus: (Total_Kotor - Nominal_Diskon) * 1.11)
df['Total_Bersih'] = (df['Total_Kotor'] - df['Nominal_Diskon']) * 1.11

print("--- TASK 2: Kolom Baru Berhasil Dibuat ---")
display(df[['id_transaksi', 'Total_Kotor', 'Nominal_Diskon', 'Total_Bersih']].head())
```

```python
# 1. Filter Multi-Kondisi
filtered_multi = df[
    (df['status'] == 'SELESAI') & 
    (df['kode_gerai'].isin(['G05', 'G09'])) & 
    (df['Total_Bersih'] > 100000)
]
print(f"Jumlah baris yang memenuhi filter multi-kondisi: {len(filtered_multi)}")

# 2. Filter dengan Query (gerai G10 dan jumlah minimal 5 unit)
filtered_query = df.query("kode_gerai == 'G10' and jumlah >= 5")
print(f"Jumlah baris gerai G10 dengan jumlah >= 5: {len(filtered_query)}")
display(filtered_query.head(3))
```

```python
# 1. Multi-Column Sorting (kode_gerai A-Z, Total_Bersih tertinggi ke terendah)
sorted_multi = df.sort_values(by=['kode_gerai', 'Total_Bersih'], ascending=[True, False])
print("--- TASK 4.1: Multi-Column Sorting (Top 3) ---")
display(sorted_multi[['id_transaksi', 'kode_gerai', 'Total_Bersih']].head(3))

# 2. Top-N Records (5 transaksi dengan Total_Bersih tertinggi)
top_5_transaksi = df.nlargest(5, 'Total_Bersih')
print("--- TASK 4.2: Top 5 Transaksi Tertinggi ---")
display(top_5_transaksi[['id_transaksi', 'kode_gerai', 'Total_Bersih']])
```

```python
# 1. Laporan Agregasi Per Gerai (Status SELESAI)
df_selesai = df[df['status'] == 'SELESAI']

laporan_gerai = df_selesai.groupby('kode_gerai').agg(
    Jumlah_Transaksi=('id_transaksi', 'count'),
    Total_Unit_Terjual=('jumlah', 'sum'),
    Total_Omzet_Bersih=('Total_Bersih', 'sum'),
    Rata_Rata_Transaksi=('Total_Bersih', 'mean')
).sort_values(by='Total_Omzet_Bersih', ascending=False).round(2)

print("--- TASK 5.1: Laporan Agregasi Per Gerai ---")
display(laporan_gerai)

# 2. Transformasi Kelompok (Menghitung Kontribusi Gerai Persen)
total_omzet_per_gerai = df.groupby('kode_gerai')['Total_Bersih'].transform('sum')
df['Kontribusi_Gerai_Persen'] = (df['Total_Bersih'] / total_omzet_per_gerai) * 100

print("--- TASK 5.2: Kolom Kontribusi_Gerai_Persen Berhasil Ditambahkan ---")
display(df[['id_transaksi', 'kode_gerai', 'Total_Bersih', 'Kontribusi_Gerai_Persen']].head())
```
