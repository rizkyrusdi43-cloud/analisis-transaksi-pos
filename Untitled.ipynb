```python
import pandas as pd

# 1. Membaca file dataset
df = pd.read_csv('paket_a_transaksi_pos_2024_2025.csv')

# 2. Menanggulangi missing value pada kolom kode_kasir
df['kode_kasir'] = df['kode_kasir'].fillna('Unknown')

# 3. Membersihkan kolom harga_satuan (mengubah koma menjadi titik, lalu ubah ke float)
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace(',', '.').astype(float)

# 4. Mengonversi kolom tanggal dan mengekstrak tahun
df['tanggal'] = pd.to_datetime(df['tanggal'], format='mixed')
df['tahun'] = df['tanggal'].dt.year

# 5. Menghitung pendapatan bersih per baris (memperhatikan diskon)
df['pendapatan'] = df['jumlah'] * df['harga_satuan'] * (1 - df['diskon_persen'] / 100)

# 6. Memfilter status transaksi (HANYA status 'SELESAI' yang dihitung)
df_selesai = df[df['status'] == 'SELESAI']

# 7. Menghitung total pendapatan per tahun (2024 dan 2025)
pendapatan_per_tahun = df_selesai.groupby('tahun')['pendapatan'].sum()

print("--- TOTAL PENDAPATAN PER TAHUN (2024 & 2025) ---")
print(pendapatan_per_tahun)
```
